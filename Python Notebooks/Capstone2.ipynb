# Data Cleaning & Preprocessing

```python
import pandas as pd
```

```python
df = pd.read_csv('movies_data.csv')
df.head()
```

```python
df.columns
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# Missing values
df.isnull().sum()
```

```python
df['certification'] = df['certification'].fillna('Unknown')
df['director'] = df['director'].fillna('Unknown')
```

```python
# Transforming the Certification Column
rating_map = {
    # Family Friendly
    'G': 'Family Friendly',
    'U': 'Family Friendly',
    '0': 'Family Friendly',
    '0+': 'Family Friendly',
    'ALL': 'Family Friendly',
    'L': 'Family Friendly',
    'B': 'Family Friendly',
    'C': 'Family Friendly',
    'AL': 'Family Friendly',
    'Genel İzleyici Kitlesi': 'Family Friendly',
    'PG': 'Family Friendly',
    '6': 'Family Friendly',
    '6+': 'Family Friendly',
    '7': 'Family Friendly',
    '8': 'Family Friendly',
    '9': 'Family Friendly',
    '10': 'Family Friendly',
    '12': 'Family Friendly',
    '12+': 'Family Friendly',
    '12A': 'Family Friendly',
    'PG12': 'Family Friendly',
    'M/12': 'Family Friendly',

    # Teen
    'PG-13': 'Teen',
    'PG13': 'Teen',
    '13': 'Teen',
    '13+': 'Teen',
    '13A': 'Teen',
    'U/A 13+': 'Teen',
    '14': 'Teen',
    '14+': 'Teen',
    '14A': 'Teen',
    '15': 'Teen',
    '15+': 'Teen',
    'R15+': 'Teen',
    'B-15': 'Teen',
    'T': 'Teen',
    'MA 15+': 'Teen',

    # Adult
    'R': 'Adult',
    '18': 'Adult',
    '18+': 'Adult',
    '18A': 'Adult',
    'R 18+': 'Adult',
    'NC-17': 'Adult',
    '16': 'Adult',
    '16+': 'Adult',
    '19': 'Adult',
    'M18': 'Adult',
    'S': 'Adult',
    'IIB': 'Adult',

    # Unknown
    'NR': 'Unknown',
    'Unknown': 'Unknown',
    'A': 'Unknown',
    'Btl': 'Unknown',
    'TP': 'Unknown',
    'M': 'Unknown',
    'APTA': 'Unknown',
    'UA': 'Unknown',
    'Ai': 'Unknown'
}

df['age_rating'] = df['certification'].map(rating_map)
```

```python
df['age_rating'].unique()
```

```python
df.drop(columns=['certification'], inplace=True)
```

```python
# Transforming the Awards column
df['awarded'] = df['awards'].notna().astype(int)
df.drop(columns=['awards'], inplace=True)
```

```python
df['awarded'].value_counts()
```

```python
df.columns
```

```python
df['genres'] = df['genres'].str.replace(' and ',', ').str.replace(',,',',')
```

```python
# New column Primary Genre of the Movie
df['primary_genre'] = df['genres'].str.split(', ').str[0].str.strip()
df['primary_genre'].unique()
```

```python
# Transforming the runtime to runtime in minutes
def convert_runtime(runtime):
    hours = 0
    minutes = 0

    if 'h' in runtime:
        hours = int(runtime.split('h')[0].strip())

    if 'm' in runtime:
        minutes_part = runtime.split('h')[-1]
        minutes = int(minutes_part.replace('m', '').strip())

    return hours * 60 + minutes

df['runtime_minutes'] = df['runtime'].apply(convert_runtime)
df.drop(columns=['runtime'], inplace=True)
```

```python
# Standardizing the budget and revenue column
import numpy as np

df[['budget', 'revenue']] = df[['budget', 'revenue']].replace('-', np.nan)
```

```python
df[['budget', 'revenue']] = df[['budget', 'revenue']].apply(
    pd.to_numeric, errors='coerce'
)
```

```python
df.loc[df['budget'] < 100, 'budget'] = np.nan
```

```python
# Duplicate rows
df.duplicated().sum()
```

```python
# Same titles
df.duplicated(subset='title').sum()
```

```python
# Same titles but different director and year
df[df.duplicated(subset='title', keep=False)]
```

```python
# New Features - Profit, ROI, Genre Count
df['profit'] = df['revenue'] - df['budget']
df['roi'] = (df['revenue'] - df['budget'])*100 / df['budget']
df['genre_count'] = df['genres'].str.count(',') + 1
```
