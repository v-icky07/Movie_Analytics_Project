#  Web Scraping Movies Data

```python
# 1st layer
import requests
```

```python
base_url = 'https://www.themoviedb.org/movie/top-rated'
```

```python
from bs4 import BeautifulSoup
```

```python
# We are getting 2000 movie links
movie_links = []
for page in range(1, 101):

    url = f"{base_url}?page={page}"

    response = requests.get(url)

    print(f"Scraping page {page}...")

    soup = BeautifulSoup(response.text, "html.parser")

    for j in soup.find_all('a',attrs={'class':"font-normal hover:underline hover:underline-offset-2 hover:decoration-current hover:text-tmdb-light-blue"}):
      s = 'https://www.themoviedb.org'+j.get('href')
      movie_links.append(s)
```

## Same web scraping procedure for 2000 movies (Looping)

```python
import time

headers = {
     "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Connection": "keep-alive",
    "Upgrade-Insecure-Requests": "1"
}

for i, url in enumerate(movie_links):

    try:
        fly = requests.get(url, headers=headers, timeout=15)

        if fly.status_code != 200:
            print(f"Failed: {i+1} | Status: {fly.status_code}")
            continue

        data = BeautifulSoup(fly.text, 'html.parser')

        # ---------------- TITLE ----------------
        title_tag = data.find('h2')

        if title_tag and title_tag.find('a'):
            c1 = title_tag.find('a').get_text(strip=True)
        else:
            c1 = None

        titles.append(c1)


        # ---------------- YEAR ----------------
        if title_tag and title_tag.find('span'):
            c2 = title_tag.find('span').get_text(strip=True)
            c2 = c2.replace('(', '').replace(')', '')
        else:
            c2 = None

        years.append(c2)


        # ---------------- CERTIFICATION ----------------
        tag = data.find('span', attrs={'class': 'certification'})

        if tag:
            c3 = tag.get_text(strip=True)
        else:
            c3 = None

        certifications.append(c3)


        # ---------------- GENRES ----------------
        tag = data.find('span', attrs={'class': 'genres'})

        if tag:
            c4 = tag.get_text(" ", strip=True)
        else:
            c4 = None

        genres.append(c4)


        # ---------------- RUNTIME ----------------
        tag = data.find('span', attrs={'class': 'runtime'})

        if tag:
            c5 = tag.get_text(" ", strip=True)
        else:
            c5 = None

        runtimes.append(c5)


        # ---------------- DIRECTOR ----------------
        profile = data.find('li', attrs={'class': 'profile'})

        if profile and profile.find('a'):
            c6 = profile.find('a').get_text(strip=True)
        else:
            c6 = None

        directors.append(c6)


        # ---------------- AWARDS ----------------
        tag = data.find(
            'div',
            attrs={
                'class': 'pr-2 text-white border-r border-white sm:pr-5 sm:text-xl'
            }
        )

        if tag:
            c7 = tag.get_text(strip=True)
        else:
            c7 = None

        awards.append(c7)


        # ---------------- FACTS ----------------
        facts = data.find('section', attrs={'class': 'facts left_column'})

        original_language = None
        budget = None
        revenue = None

        if facts:

            for p in facts.find_all('p'):

                bdi = p.find('bdi')

                if not bdi:
                    continue

                label = bdi.get_text(strip=True)
                value = p.get_text(" ", strip=True)

                value = value.replace(label, '', 1).strip()

                if label == 'Original Language':
                    original_language = value

                elif label == 'Budget':
                    budget = value.replace('$', '').replace(',', '')

                elif label == 'Revenue':
                    revenue = value.replace('$', '').replace(',', '')


        original_languages.append(original_language)
        budgets.append(budget)
        revenues.append(revenue)


        # ---------------- RATING ----------------
        tag = data.find('div', attrs={'class': 'user_score_chart'})

        if tag:
            c11 = tag.get('data-percent')
        else:
            c11 = None

        ratings.append(c11)


        print(f"{i+1}/{len(movie_links)} - {c1}")

        # Small delay
        time.sleep(0.5)


    except Exception as e:

        print(f"ERROR: {i+1} | {url}")
        print(e)

        # Keep lists aligned
        titles.append(None)
        years.append(None)
        certifications.append(None)
        genres.append(None)
        runtimes.append(None)
        directors.append(None)
        awards.append(None)
        original_languages.append(None)
        budgets.append(None)
        revenues.append(None)
        ratings.append(None)
```

## Creating a Pandas Dataframe from the scraped data

```python
import pandas as pd
df = pd.DataFrame({
    'title': titles,
    'year': years,
    'certification': certifications,
    'genres': genres,
    'runtime': runtimes,
    'director': directors,
    'awards': awards,
    'original_language': original_languages,
    'budget': budgets,
    'revenue': revenues,
    'rating': ratings
})
```
